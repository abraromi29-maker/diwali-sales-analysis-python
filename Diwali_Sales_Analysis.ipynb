**Customer Order Analysis Across Different Demographic Segments**

```python
df.columns
```

```python
gender=df.groupby(['Gender'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False)
age=df.groupby(['Age Group'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False)
marit=df.groupby(['Marital_Status'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False)
state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(5)
occupation=df.groupby(['Occupation'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(5)
category=df.groupby(['Product_Category'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(4)
```

```python

# Create subplot layout
fig, axes = plt.subplots(2, 3, figsize=(20,10))

sns.barplot(x='Gender', y='Orders', data=gender, ax=axes[0,0])
sns.barplot(x='Age Group', y='Orders', data=age, ax=axes[0,1])
sns.barplot(x='State', y='Orders', data=state, ax=axes[0,2])

sns.barplot(x='Product_Category', y='Orders', data=category, ax=axes[1,0])
sns.barplot(x='Occupation', y='Orders', data=occupation, ax=axes[1,1])
sns.barplot(x='Marital_Status', y='Orders', data=marit, ax=axes[1,2])

plt.tight_layout()
plt.show()
```

The analysis shows that:<br>

1.**Female customers** placed the highest number of orders<br>
2.**The 26–35 age group** is the most active customer segment<br>
3.**Uttar Pradesh, Maharashtra, and Karnataka** are the strongest markets<br>
4.**Clothing & Apparel** is the most ordered product category<br>
5.**IT sector professionals** contributed the highest number of orders<br>
6.**Married customers** showed relatively higher purchasing behavior<br>

**Final Business Conclusion:**<br>
**During the Diwali sales period,The most valuable target customer segment appears to be female working professionals aged 26–35 from high-performing states, with strong interest in Clothing & Apparel products. Focusing marketing, inventory planning, and customer engagement on this segment could significantly improve business growth.**
